$$ \huge\textbf{Model Development}$$

```python
laptop_data.columns
```

```python
y = laptop_data['Price']
X = laptop_data[['Processor_Speed', 'RAM_Size', 'Storage_Capacity']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test, y_train  , y_test = train_test_split(X , y , test_size = .2)
```

```python
print(len(X_train))
print(len(y_train))
```

```python
from sklearn.ensemble import RandomForestRegressor

rfrmodel = RandomForestRegressor()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_gridrfr = {'max_depth':[2,5,10,15,20,25] , 
                 'max_features': ['auto' , 'log2'] , 
                 'n_estimators': [2,4,6,8,10,12]}

gridrfr = GridSearchCV(rfrmodel , param_gridrfr)
```

```python
gridrfr.fit(X_train , y_train)
```

```python
gridrfr.best_params_
```

```python
prediction = gridrfr.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mean_absolute_error(y_test , prediction )
```

```python
joblib.dump(gridrfr , 'rf_model.pkl')
```

$$\huge{Loan\ Approval}$$

```python
loan_data = pd.read_csv('Loan.csv')
loan_data.head()
```

```python
print(loan_data.columns.tolist())
```

```python
columns_selection = ['Age','CreditScore','MonthlyDebtPayments','MonthlyIncome','MonthlyLoanPayment','RiskScore']

X = loan_data.loc[: , columns_selection]
```

```python
X.corr(method='pearson')
```

```python
y = loan_data['LoanAmount']
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size=.2)
```

```python
print(len(X_train))
print(len(y_train))
```

```python
from sklearn.ensemble import RandomForestRegressor

ranfreg = RandomForestRegressor()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_gridrfr = {'max_depth':[2,5,10,15,20,25] , 
                 'max_features': ['auto' , 'log2'] , 
                 'n_estimators': [2,4,6,8,10,12]}

gridrfr = GridSearchCV(ranfreg , param_gridrfr)
```

```python
gridrfr.fit(X_train , y_train)
```

```python
gridrfr.best_params_
```

```python
prediction = gridrfr.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mean_absolute_error(y_test , prediction)
```

```python
print(X.columns.tolist())
```

```python
joblib.dump(gridrfr , 'randomfReg.pkl')
```
